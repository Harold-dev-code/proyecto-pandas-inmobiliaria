##Desafío: vamos a practicar

El equipo de ML ha llegado con algunas demandas de última hora que debemos resolver en este momento de análisis exploratorio. Estas demandas son las siguientes:



1.   Calcular el promedio de habitaciones por departamento.
2.   Verificar cuántas colonias únicas existen en nuestra base de datos
3.   Analizar qué colonias tienen el promedio de alquiler más alto.
4.   Crear un gráfico de barras horizontales que muestre las 5 colonias con l  os  promedios de alquiler más altos.

```
# 1. Calcular el promedio de habitaciones por departamento.
df.groupby('Tipo')[['Habitaciones']].mean()
```

```
#segunda manera para Calcular el promedio de habitaciones por departamento.
df['Habitaciones'].mean()
```

```
#2.Verificar cuántas colonias únicas existen en nuestra base de datos
df['Colonia'].unique()
#Una forma de verificar cuántos barrios únicos existen en la base de datos sería utilizando el método unique(), que nos devolverá una lista con los barrios únicos.
```

```
#usamos la función len() para contar cuántos elementos hay en esa lista:
len(df['Colonia'].unique())
```

```
# Además, también podemos utilizar un método de Pandas que realiza directamente el conteo de elementos únicos en una columna específica. Este método es nunique():
df['Colonia'].nunique()
```

```
#3. Analizar qué colonias tienen el promedio de alquiler más alto.
df.groupby('Colonia')[['Valor']].mean().sort_values('Valor', ascending=False)
'''Para analizar qué barrios tienen el promedio de valor de alquiler más alto, podemos:

-Agrupar nuestra base de datos por la columna "Colonia".
-Aplicar el método mean.
-Colocar el resultado en un DataFrame y ordenarlo.'''
```

```
# 4. Crear un gráfico de barras horizontales que muestre las 5 colonias con l os promedios de alquiler más altos.
df_colonias = df.groupby('Colonia')[['Valor']].mean().sort_values('Valor', ascending=False).head()
'''Como queremos crear un gráfico con solo las 5 colonias que tienen el promedio de alquiler más alto, podemos usar .head() para seleccionar las 5 primeras filas del DataFrame que creamos en el desafío 3 y guardar eso en una variable:'''
```

```
#Luego, podemos crear el gráfico:
df_colonias.plot(kind='bar', figsize=(14,10), color='blue');
```

##Tratar valores nulos


Dado que nuestros datos serán utilizados por el equipo de ML para crear un modelo para estimar el precio de alquiler de las propiedades, es importante que tratemos los datos faltantes, ya que los datos nulos no pueden utilizarse en el entrenamiento de modelos de ML.

###Verificar la existencia de datos nulos

```
df.isnull()
```

```
df.isnull().sum() # cuando trato con booleanos yo puedo sumar y esta operacion me devuelve cada columna con la cantidad de nulos que tiene
```

El equipo de ML nos dan la instruccion que los datos nulos le asignemos cero.

```
df.fillna(0) # esto llena los valores nulos y lo que este entre parantesis , en este caso cero
```

```
df= df.fillna(0) #actualizo el dataframe con el cambio
df.isnull().sum() # verifico que no tenga datos nulos
```
